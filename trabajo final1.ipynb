```python
# ============================================
# TALLER FINAL – LIMPIEZA Y PREPARACIÓN DE DATOS
# Maria Paz Clavijo, Mariana Moreno, René Quintana
# ============================================

# --------------------------------------------
# 1. Importación de librerías
# --------------------------------------------
import pandas as pd
import numpy as np
import unidecode

# --------------------------------------------
# 2. Funciones de limpieza
# --------------------------------------------

def limpiar_columnas(df):
    """
    Limpia nombres de columnas eliminando tildes y caracteres dañados.
    """
    nuevas = []
    for col in df.columns:
        col = unidecode.unidecode(col)
        col = col.replace("?", "Ñ")
        nuevas.append(col)
    df.columns = nuevas
    return df


def limpiar_numeros(df, excepciones=[]):
    """
    Limpia columnas numéricas:
    - elimina comas, espacios, comillas
    - convierte a número
    """
    for col in df.columns:
        if col in excepciones:
            continue
        df[col] = df[col].astype(str)
        df[col] = (
            df[col]
            .str.replace(",", "")
            .str.replace('"', "")
            .str.replace(" ", "")
        )
        df[col] = pd.to_numeric(df[col], errors="ignore")
    return df


# --------------------------------------------
# 3. Carga de archivos (solo los 3 solicitados)
# --------------------------------------------

empresas = pd.read_csv("empresas_10000.csv", encoding="latin1")
poblacion = pd.read_csv("poblacion_sisben.csv", encoding="latin1")
establecimientos = pd.read_csv("establecimientos_comercio_2018.csv", encoding="latin1")

print("✔ Archivos cargados correctamente")


# --------------------------------------------
# 4. Limpieza de encabezados
# --------------------------------------------

empresas = limpiar_columnas(empresas)
poblacion = limpiar_columnas(poblacion)
establecimientos = limpiar_columnas(establecimientos)

print("✔ Encabezados limpiados")


# --------------------------------------------
# 5. Limpieza específica para POBLACIÓN SISBÉN
# --------------------------------------------

# Suponemos que "Sectores" NO es numérico
columnas_num_pob = poblacion.columns.drop("Sectores")

for col in columnas_num_pob:
    poblacion[col] = (
        poblacion[col]
        .astype(str)
        .str.replace(",", "")
        .str.replace('"', "")
        .str.replace(" ", "")
    )
    poblacion[col] = pd.to_numeric(poblacion[col], errors="coerce")

# Añadir depto/municipio
poblacion["departamento"] = "Antioquia"
poblacion["municipio"] = "La Estrella"

# Reordenar
poblacion = poblacion[
    ["departamento", "municipio", "Sectores"] + list(columnas_num_pob)
]

print("✔ Población SISBÉN limpiada correctamente")


# --------------------------------------------
# 6. Limpieza general para las otras bases
# --------------------------------------------

empresas = limpiar_numeros(empresas)
establecimientos = limpiar_numeros(establecimientos)

print("✔ Limpieza general aplicada")


# --------------------------------------------
# 7. Agregar departamento y municipio si faltan
# --------------------------------------------

bases_geo = [empresas, establecimientos]

for df in bases_geo:
    if "departamento" not in df.columns:
        df["departamento"] = "Antioquia"
    if "municipio" not in df.columns:
        df["municipio"] = "La Estrella"

print("✔ Columnas geográficas agregadas")


# --------------------------------------------
# 8. Exportación de archivos limpios
# --------------------------------------------

empresas.to_csv("empresas_limpio.csv", index=False, encoding="utf-8")
poblacion.to_csv("poblacion_sisben_limpio.csv", index=False, encoding="utf-8")
establecimientos.to_csv("establecimientos_limpio.csv", index=False, encoding="utf-8")

print("🎉 LISTO: Las 3 bases limpias fueron exportadas correctamente.")
```
